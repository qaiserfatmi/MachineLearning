# 3. Reshaping, Arranging, Raveling, Transposing Array

```python
# arange function -> range print array
# start,end,steps

arr_1D = np.arange(0,15,2)
print(arr_1D)
```

```python
# reshape function -> Reshape dim of array  
# 3D,row,columns

arr_res = arr_1D.reshape(2,4)
print(arr_res)
```

```python
arr_res = arr_1D.reshape(2,2,2)
print(arr_res)
```

```python
# ravel function -> conv nD array -> 1D array
# flatten function
arr_conv = arr_res.ravel()
print(arr_conv)
```

```python
# transpose matrix
# row -> col -> row

arr_res = arr_1D.reshape(2,4)
print(arr_res)

arr_res.transpose()
# or
arr_res.T
```

### 4.6. Max, Min and Numerical OPS

# 5. Slicing

```python
arr_01 = np.arange(1,101).reshape(10,10)
print(arr_01)
```

```python
# Get Element = [row,column] ; retrieves the element located in the first row and fourth column 
arr_01[0,3]
```

```python
arr_01[ : , : ]    # Selects whole row and whole column
```

```python
arr_01[ 3 , : ]    # Selects whole row 3
```

```python
arr_01[ : , 4 ]    # Selects whole column 4
```

```python
arr_01[0:2 , 0]    # arr[row , column]
                   # 0:2 → slice from row 0 to row 1 (Python slices exclude the end index)
                   # 0 → refers to column 0 (the first column)
```

```python
arr_01[2 , 1:4]    # arr[row , column]
                   # 2 → refers to row 2 (the third row)
                   # 1:4 → slice from column 1 to column 3 (Python slices exclude the end index)
```

```python
arr_01[: , 0:4 ]   # : referes all rows and slice columns from 0-3 (exclude last index)
```

```python
arr_01[0,5]    # retrieves the element located in the first row and sixth column
```

```python
arr_01[0,5].ndim    # arr_01[0, 5] is a scalar value (e.g., an integer or float)
                	# A scalar in NumPy has 0 dimensions
```

```python
arr_01[: , 0:4]     # retrieves all rows and columns from 0-3 (excluding last index)
```

```python
arr_01[: , 0:4].ndim
```

# 6. Concatenation

```python
arr_01 = np.arange(1,17).reshape(4,4)
arr_02 = np.arange(17,33).reshape(4,4)
print(arr_01)
print(arr_02)
```

```python
arr_01 + arr_02   # Simple addition
```

```python
np.concatenate((arr_01, arr_02))   # Joining two arrays
```

```python
np.vstack((arr_01, arr_02))   # Joining two arrays vertically
```

```python
np.vstack((arr_01,arr_02))   # Joining two arrays vertically
```

```python
np.hstack((arr_01, arr_02))   # Joining two arrays horizontally
```

```python
np.concatenate((arr_01, arr_02), axis=1)   # join two arrays side-by-side along columns (i.e., horizontally)
```

```python
np.concatenate((arr_01, arr_02), axis=0)   # Join two arrays through stacking  (i.e., vertically)
```

```python
x_axis = np.arange(0,3*np.pi,0.1)  
# creates a NumPy array of values from 0 to just under 3\pi (approximately 9.42), in steps of 0.1.

print (x_axis)
```

```python
y_axis = np.sin(x_axis)   # computes the sine of each value in the x_axis array.

print (y_axis)
```

### 8.1. Plotting

```python
import matplotlib.pyplot as plt   # for plotting trigonometric functions like sine or cosine
plt.plot(x_axis,y_axis)
plt.title("Sine Wave")
plt.xlabel("x (radians)")
plt.ylabel("sin(x)")
plt.grid(True)
plt.show()
```

# 9. Random Sampling

```python
import random
```

```python
np.random.random(1)   # generates one random float between 0.0 and 1.0 (excluding 1)
```

```python
np.random.random((4,4))  # Creates 4×4 NumPy array filled with random float values between 0.0 and 1.0
```

```python
np.random.randint(1,9,(4,4))  # (start, end, (Rows, Column))
```

```python
np.random.randint(1,9,(2,6,4)) # (start, end, (Blocks , Rows, Columns))   ; end is non-inclusive
# Generates a 3D NumPy array with shape (2, 6, 4) filled with random integers from 1 (inclusive) to 9 (exclusive).
```

```python
np.random.rand(1,3,4,4)  # (Batch, Block, Rows, Column)
# creates a 4D NumPy array with shape (1, 2, 3, 4) filled with random float values between 0.0 and 1.0
```

```python
np.random.rand(2,2,3,5)  # (Batch, Block, Rows, Column)
```

```python
list = [1,2,3,4,5,6]
np.random.choice(list)   # selects one random element from the list
```

```python
np.random.permutation(list)
# returns a randomly shuffled version of the input list or array — without modifying the original
```

# 10. String Operations

```python
print (myStr)
print (myStr2)
```

```python
np.char.lower(myStr)   # Lower Case
```

```python
np.char.upper(myStr2)  # Upper Case
```

```python
np.char.split(myStr2)  # Split words
```

```python
np.char.splitlines(myStr2) 
# splits a string into a list of lines at newline characters (\n, \r, or \r\n)
```

```python
np.char.equal(myStr,myStr2)  # Checks equality
```
